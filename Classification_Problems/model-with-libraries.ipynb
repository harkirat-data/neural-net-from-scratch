```python
from sklearn.datasets import make_classification
import numpy as np
import pandas as pd

import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
X, y = make_classification(
    n_samples=75,
    n_features=2,
    n_redundant=0,
    n_informative=2,
    n_clusters_per_class=1,
    random_state=42
)

cgpa = 4 + (X[:,0] - X[:,0].min()) / (X[:,0].max() - X[:,0].min()) * 6
profile_score = 40 + (X[:,1] - X[:,1].min()) / (X[:,1].max() - X[:,1].min()) * 60
```

```python
df = pd.DataFrame({
    "cgpa": cgpa,
    "profile_score": profile_score,
    "placed": y
})

df["cgpa"] = (df["cgpa"] - df["cgpa"].mean()) / df["cgpa"].std()
df["profile_score"] = (
    (df["profile_score"] - df["profile_score"].mean())
    / df["profile_score"].std()
)

df.head()
```

```python
model = Sequential()

model.add(Dense(2,activation="sigmoid",input_dim=2))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```

```python
new_weights = [
    np.array([[0.1,0.1],
              [0.1,0.1]], dtype=np.float32),

    np.array([0.,0.], dtype=np.float32),

    np.array([[0.1],
              [0.1]], dtype=np.float32),

    np.array([0.], dtype=np.float32)
]

model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.3)

model.compile(loss="binary_crossentropy",optimizer=optimizer,metrics=["accuracy"])
```

```python
history = model.fit(df.iloc[:,0:-1].values,df["placed"].values,epochs=100,batch_size=1,verbose=1)
```
